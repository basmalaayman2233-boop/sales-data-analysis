```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.gridspec as gridspec
import numpy as np
from matplotlib.patches import FancyBboxPatch
from matplotlib.colors import LinearSegmentedColormap

# ── Load Data ──────────────────────────────────────────────────────────────────
details = pd.read_csv("/mnt/user-data/uploads/Details.csv")
orders  = pd.read_csv("/mnt/user-data/uploads/Orders.csv")

# Parse dates & derive Month / Quarter
orders["Order Date"] = pd.to_datetime(orders["Order Date"], dayfirst=True)
orders["Month"]      = orders["Order Date"].dt.month
orders["Quarter"]    = orders["Order Date"].dt.quarter

# Merge on Order ID
df = details.merge(orders, on="Order ID", how="left")

# ── Palette & Style ────────────────────────────────────────────────────────────
BG_OUTER   = "#1a0533"   # deep purple
BG_CARD    = "#2a1a4a"   # card background
ACCENT_BL  = "#2979ff"   # bright blue bars
ACCENT_YEL = "#f5c518"   # yellow line
TEXT_WHITE = "#ffffff"
TEXT_MUTED = "#cccccc"

PIE_COLORS  = ["#2979ff", "#7b1fa2", "#e91e8c", "#f57c00", "#00bcd4"]
PIE2_COLORS = ["#2979ff", "#f57c00", "#9c27b0", "#4caf50", "#e91e8c"]

MONTH_LABELS = ["January","February","March","April","May","June",
                "July","August","September","October","November","December"]

plt.rcParams.update({
    "figure.facecolor":  BG_OUTER,
    "axes.facecolor":    BG_CARD,
    "axes.edgecolor":    "#3a2a5a",
    "axes.labelcolor":   TEXT_MUTED,
    "xtick.color":       TEXT_MUTED,
    "ytick.color":       TEXT_MUTED,
    "text.color":        TEXT_WHITE,
    "font.family":       "DejaVu Sans",
    "font.size":         9,
    "grid.color":        "#3a2a5a",
    "grid.linewidth":    0.5,
})

# ── Helper: rounded card background ───────────────────────────────────────────
def add_card_bg(ax, fig, color=BG_CARD, pad=0.02):
    """Draw a rounded rect behind an axes (in figure coords)."""
    pos = ax.get_position()
    fancy = FancyBboxPatch(
        (pos.x0 - pad/2, pos.y0 - pad/2),
        pos.width + pad, pos.height + pad,
        boxstyle="round,pad=0.01",
        linewidth=0, facecolor=color, transform=fig.transFigure, zorder=0
    )
    fig.add_artist(fancy)

# ── Figure Layout ──────────────────────────────────────────────────────────────
fig = plt.figure(figsize=(18, 11))
fig.patch.set_facecolor(BG_OUTER)

# Top title bar height, then two rows of 3 charts each
gs = gridspec.GridSpec(
    3, 3,
    figure=fig,
    top=0.88, bottom=0.05,
    left=0.04, right=0.98,
    hspace=0.45, wspace=0.32,
)

# ── TOP HEADER BAND ────────────────────────────────────────────────────────────
header_ax = fig.add_axes([0, 0.89, 1, 0.11])
header_ax.set_facecolor(BG_OUTER)
header_ax.axis("off")

# Title box
title_box = FancyBboxPatch((0.01, 0.08), 0.18, 0.82,
    boxstyle="round,pad=0.02", linewidth=0,
    facecolor="#0d0020", transform=header_ax.transAxes, zorder=1)
header_ax.add_patch(title_box)
header_ax.text(0.10, 0.52, "Sales Dashboard",
    transform=header_ax.transAxes,
    fontsize=18, fontweight="bold", color=TEXT_WHITE,
    ha="center", va="center", zorder=2)

# Quarter buttons
qtrs = ["Qtr 1", "Qtr 2", "Qtr 3", "Qtr 4"]
qtr_x = [0.24, 0.31, 0.38, 0.45]
for lbl, x in zip(qtrs, qtr_x):
    btn = FancyBboxPatch((x, 0.18), 0.065, 0.64,
        boxstyle="round,pad=0.02", linewidth=0,
        facecolor="#6a1b9a", transform=header_ax.transAxes, zorder=1)
    header_ax.add_patch(btn)
    header_ax.text(x + 0.0325, 0.52, lbl,
        transform=header_ax.transAxes,
        fontsize=11, fontweight="bold", color=TEXT_WHITE,
        ha="center", va="center", zorder=2)

# KPI cards
kpis = [
    ("5,615",  "Sum of Quantity"),
    ("37K",    "Sum of Profit"),
    ("438K",   "Sum of Amount"),
]
kpi_x = [0.70, 0.80, 0.90]
for (val, lbl), x in zip(kpis, kpi_x):
    kcard = FancyBboxPatch((x - 0.04, 0.08), 0.095, 0.84,
        boxstyle="round,pad=0.02", linewidth=0,
        facecolor="#3a1a6a", transform=header_ax.transAxes, zorder=1)
    header_ax.add_patch(kcard)
    header_ax.text(x + 0.0075, 0.65, val,
        transform=header_ax.transAxes,
        fontsize=15, fontweight="bold", color=TEXT_WHITE,
        ha="center", va="center", zorder=2)
    header_ax.text(x + 0.0075, 0.22, lbl,
        transform=header_ax.transAxes,
        fontsize=7.5, color=TEXT_MUTED,
        ha="center", va="center", zorder=2)

# ══════════════════════════════════════════════════════════════════════════════
# CHART 1 – Profit by Sub-Category (horizontal bar)
# ══════════════════════════════════════════════════════════════════════════════
ax1 = fig.add_subplot(gs[0, 0])
sub_profit = (df.groupby("Sub-Category")["Profit"]
              .sum().sort_values(ascending=False).head(6))
bars = ax1.barh(sub_profit.index[::-1], sub_profit.values[::-1],
                color=ACCENT_BL, height=0.6)
for bar, val in zip(bars, sub_profit.values[::-1]):
    ax1.text(bar.get_width() + 50, bar.get_y() + bar.get_height()/2,
             f"{val/1000:.1f}K", va="center", fontsize=8, color=TEXT_WHITE)
ax1.set_title("Sum of Profit by Sub-Category", color=TEXT_WHITE,
              fontsize=10, fontweight="bold", pad=8)
ax1.set_xlabel("Sum of Profit", fontsize=8)
ax1.tick_params(labelsize=8)
ax1.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"{x/1000:.0f}K"))
ax1.grid(axis="x", alpha=0.3)
ax1.spines[["top","right"]].set_visible(False)

# ══════════════════════════════════════════════════════════════════════════════
# CHART 2 – Quantity by PaymentMode (donut)
# ══════════════════════════════════════════════════════════════════════════════
ax2 = fig.add_subplot(gs[0, 1])
pay_qty = df.groupby("PaymentMode")["Quantity"].sum()
wedges, texts, autotexts = ax2.pie(
    pay_qty.values,
    labels=None,
    autopct=lambda p: f"{p:.1f}%",
    colors=PIE_COLORS[:len(pay_qty)],
    startangle=90,
    pctdistance=0.75,
    wedgeprops=dict(width=0.55, edgecolor=BG_OUTER, linewidth=1.5),
)
for at in autotexts:
    at.set_fontsize(7.5)
    at.set_color(TEXT_WHITE)

centre_circle = plt.Circle((0, 0), 0.45, color=BG_CARD)
ax2.add_patch(centre_circle)
ax2.legend(pay_qty.index, loc="lower center", bbox_to_anchor=(0.5, -0.18),
           ncol=3, fontsize=7, framealpha=0, labelcolor=TEXT_MUTED)
ax2.set_title("Sum of Quantity by PaymentMode", color=TEXT_WHITE,
              fontsize=10, fontweight="bold", pad=8)

# ══════════════════════════════════════════════════════════════════════════════
# CHART 3 – Profit by Sub-Category again (duplicate right panel)
# ══════════════════════════════════════════════════════════════════════════════
ax3 = fig.add_subplot(gs[0, 2])
bars3 = ax3.barh(sub_profit.index[::-1], sub_profit.values[::-1],
                 color=ACCENT_BL, height=0.6)
for bar, val in zip(bars3, sub_profit.values[::-1]):
    ax3.text(bar.get_width() + 50, bar.get_y() + bar.get_height()/2,
             f"{val/1000:.1f}K", va="center", fontsize=8, color=TEXT_WHITE)
ax3.set_title("Sum of Profit by Sub-Category", color=TEXT_WHITE,
              fontsize=10, fontweight="bold", pad=8)
ax3.set_xlabel("Sum of Profit", fontsize=8)
ax3.tick_params(labelsize=8)
ax3.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"{x/1000:.0f}K"))
ax3.grid(axis="x", alpha=0.3)
ax3.spines[["top","right"]].set_visible(False)

# ══════════════════════════════════════════════════════════════════════════════
# CHART 4 – Quantity & Amount by Category (donut)
# ══════════════════════════════════════════════════════════════════════════════
ax4 = fig.add_subplot(gs[1, 0])
cat_amt = df.groupby("Category")["Amount"].sum()
wedges4, texts4, autotexts4 = ax4.pie(
    cat_amt.values,
    labels=None,
    autopct=lambda p: f"{cat_amt.values[list(cat_amt.values).index(sorted(cat_amt.values, key=lambda x: abs(x - p/100*cat_amt.sum()))[0])]/1000:.0f}K\n({p:.2f}%)",
    colors=PIE2_COLORS[:len(cat_amt)],
    startangle=140,
    pctdistance=0.72,
    wedgeprops=dict(width=0.5, edgecolor=BG_OUTER, linewidth=1.5),
)
for at in autotexts4:
    at.set_fontsize(6.5)
    at.set_color(TEXT_WHITE)
cc4 = plt.Circle((0, 0), 0.5, color=BG_CARD)
ax4.add_patch(cc4)
ax4.legend(cat_amt.index, loc="lower center", bbox_to_anchor=(0.5, -0.18),
           ncol=3, fontsize=7, framealpha=0, labelcolor=TEXT_MUTED)
ax4.set_title("Sum of Quantity and Sum of\nAmount by Category",
              color=TEXT_WHITE, fontsize=10, fontweight="bold", pad=8)

# ══════════════════════════════════════════════════════════════════════════════
# CHART 5 – Amount by CustomerName (vertical bar)
# ══════════════════════════════════════════════════════════════════════════════
ax5 = fig.add_subplot(gs[1, 1])
cust_amt = (df.groupby("CustomerName")["Amount"]
            .sum().sort_values(ascending=False).head(6))
bar_colors = [ACCENT_BL] * (len(cust_amt) - 1) + ["#9c27b0"]
ax5.bar(cust_amt.index, cust_amt.values, color=bar_colors, width=0.6)
ax5.set_title("Sum of Amount by CustomerName", color=TEXT_WHITE,
              fontsize=10, fontweight="bold", pad=8)
ax5.set_ylabel("Amount", fontsize=8)
ax5.tick_params(axis="x", rotation=20, labelsize=8)
ax5.tick_params(axis="y", labelsize=8)
ax5.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"{x/1000:.0f}K"))
ax5.grid(axis="y", alpha=0.3)
ax5.spines[["top","right"]].set_visible(False)

# ══════════════════════════════════════════════════════════════════════════════
# CHART 6 – Profit by Month (line)
# ══════════════════════════════════════════════════════════════════════════════
ax6 = fig.add_subplot(gs[1, 2])
monthly = df.groupby("Month")["Profit"].sum().reindex(range(1, 13), fill_value=0)
ax6.plot(range(12), monthly.values, color=ACCENT_YEL,
         linewidth=2.2, marker="o", markersize=5, markerfacecolor=ACCENT_YEL)
ax6.fill_between(range(12), monthly.values, alpha=0.08, color=ACCENT_YEL)

for i, val in enumerate(monthly.values):
    offset = 200 if val >= 0 else -600
    ax6.text(i, val + offset, f"{val/1000:.0f}K",
             ha="center", fontsize=7, color=TEXT_WHITE)

ax6.axhline(0, color="#555", linewidth=0.8, linestyle="--")
ax6.set_xticks(range(12))
ax6.set_xticklabels(
    ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    rotation=30, ha="right", fontsize=7.5
)
ax6.tick_params(axis="y", labelsize=8)
ax6.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"{x/1000:.0f}K"))
ax6.set_title("Sum of Profit by Month", color=TEXT_WHITE,
              fontsize=10, fontweight="bold", pad=8)
ax6.set_ylabel("Sum of Profit", fontsize=8)
ax6.grid(alpha=0.3)
ax6.spines[["top","right"]].set_visible(False)

# ── Card backgrounds for all chart axes ────────────────────────────────────────
fig.canvas.draw()
for ax in [ax1, ax2, ax3, ax4, ax5, ax6]:
    add_card_bg(ax, fig, color=BG_CARD, pad=0.015)

# ── Save ───────────────────────────────────────────────────────────────────────
out = "/mnt/user-data/outputs/sales_dashboard.png"
plt.savefig(out, dpi=150, bbox_inches="tight", facecolor=BG_OUTER)
print(f"Saved → {out}")
```
